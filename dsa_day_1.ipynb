```python
n=list(map(int,input("enter value:").split()))
k=int(input("enter value of k:"))
arr=(n)
a=k
max=0
for i in range (0,len(arr)-a+1):
    total=0
    for j in range(i,i+k):
        total+=arr[j]
        if total>max:
         max=total
print(max)
```

```python
#longest subarray
s = input("Enter the string: ")
k = int(input("Enter the value of K: "))

n = len(s)
large= 0

for i in range(n):
    count = set()

    for j in range(i, n):
        count.add(s[j])

        if len(count) > k:
            break

        current_length = j - i + 1

        if current_length > large:
            large = current_length
print(large)
```
